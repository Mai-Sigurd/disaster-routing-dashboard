# Fix trip stats

```python
import pandas as pd
df = pd.read_csv('Amager_60min_geopop_div3_basis/dijkstra-fastest-path-output/analysis/analysis/trip_stats_disaster.csv')
df = df.rename(columns={'car': 'Basis'})
df.to_csv('trip_stats_disaster_basis.csv', index=False)

def fix_trip_stats(folder_name: str, column_to: str, name, original_df: str = 'trip_stats_disaster_basis.csv', remove_basis: bool = False):
    df = pd.read_csv(original_df)
    new_df = pd.read_csv(f'{folder_name}/analysis/analysis/trip_stats_disaster.csv')
    df[column_to] = new_df['car']
    if remove_basis:
        df.drop(columns=['Basis'], inplace=True)
    df.to_csv(f'trip_stats_disaster_{name}.csv', index=False)
```

```python
## POPULATION
fix_trip_stats('Amager_60min_number_div3/dijkstra-fastest-path-output/','Number', 'population')
```

```python
# MORE EXITS
fix_trip_stats('AmagerMoreExits_60min_geopop_div3/dijkstra-fastest-path-output/','More Exits', 'moreExits')
```

```python
# TIME
fix_trip_stats('Amager_90min_geopop_div3/dijkstra-fastest-path-output/','90 min', 'time')
fix_trip_stats('Amager_120min_geopop_div3/dijkstra-fastest-path-output/','120 min', 'time','trip_stats_disaster_time.csv')
fix_trip_stats('Amager_150min_geopop_div3/dijkstra-fastest-path-output/','150 min', 'time','trip_stats_disaster_time.csv')
fix_trip_stats('Amager_180min_geopop_div3/dijkstra-fastest-path-output/','180 min', 'time','trip_stats_disaster_time.csv')
```

```python
#ALGORITHMS
fix_trip_stats('fastest-shortest-combined/amager-fastest-3-output','Fastest', 'algorithms', remove_basis=True)
fix_trip_stats('fastest-shortest-combined/amager-shortest-3-output','Shortest', 'algorithms', 'trip_stats_disaster_algorithms.csv')
fix_trip_stats('polaris-combined/amager-polaris-3-output','Polaris', 'algorithms', 'trip_stats_disaster_algorithms.csv')
```

```python
## DIVERSIFYING ENDPOINTS
fix_trip_stats('fastest-shortest-combined/amager-fastest-1-output','Fastest - div:1', 'diversifying', remove_basis=True)
fix_trip_stats('fastest-shortest-combined/amager-shortest-1-output','Shortest - div:1', 'diversifying', 'trip_stats_disaster_diversifying.csv')
fix_trip_stats('polaris-combined/amager-polaris-1-output','Polaris - div:1', 'diversifying', 'trip_stats_disaster_diversifying.csv')
# div 3
fix_trip_stats('fastest-shortest-combined/amager-fastest-3-output','Fastest - div:3', 'diversifying','trip_stats_disaster_diversifying.csv')
fix_trip_stats('fastest-shortest-combined/amager-shortest-3-output','Shortest - div:3', 'diversifying', 'trip_stats_disaster_diversifying.csv')
fix_trip_stats('polaris-combined/amager-polaris-3-output','Polaris - div:3', 'diversifying', 'trip_stats_disaster_diversifying.csv')
```

```python
# Ravenna
fix_trip_stats('fastest-shortest-combined/ravenna-fastest-3-output','Fastest', 'ravenna', remove_basis=True)
fix_trip_stats('fastest-shortest-combined/ravenna-shortest-3-output','Shortest', 'ravenna', 'trip_stats_disaster_ravenna.csv')
fix_trip_stats('polaris-combined/ravenna-polaris-3-output','Polaris', 'ravenna', 'trip_stats_disaster_ravenna.csv')
```

# Fix dangerzone stats

```python
value = ' Value'
import pandas as pd
df = pd.read_csv('Amager_60min_geopop_div3_basis/dijkstra-fastest-path-output/analysis/danger_zone_data.csv')
df = df.rename(columns={value: 'Basis'})
df.to_csv('danger_zone_data_basis.csv', index=False)



def fix_danger_zone_data(folder_name: str, column_to: str, name, original_df: str = 'danger_zone_data_basis.csv', rename: str = "", remove_basis: bool = False):
    basis_df = pd.read_csv(original_df)
    new_df = pd.read_csv(f'{folder_name}/analysis/danger_zone_data.csv')
    basis_df[column_to] = new_df[value]
    if rename != "":
        basis_df.rename(columns={'Basis': rename}, inplace=True)
    if remove_basis:
        basis_df.drop(columns=['Basis'], inplace=True)
    basis_df.to_csv(f'danger_zone_data_{name}.csv', index=False)
```

```python
## POPULATION
fix_danger_zone_data(folder_name='Amager_60min_geopop_div3_basis/dijkstra-fastest-path-output/',column_to='Number', name='population')
```

```python
## MORE EXITS
fix_danger_zone_data(folder_name='AmagerMoreExits_60min_geopop_div3/dijkstra-fastest-path-output/',column_to='More Exits', name='moreExits')
```

```python
#TIME
fix_danger_zone_data(folder_name='Amager_90min_geopop_div3/dijkstra-fastest-path-output/',column_to='90 min', name='time', rename='Basis - 60 min')
fix_danger_zone_data('Amager_120min_geopop_div3/dijkstra-fastest-path-output/','120 min', 'time','danger_zone_data_time.csv')
fix_danger_zone_data('Amager_150min_geopop_div3/dijkstra-fastest-path-output/','150 min', 'time','danger_zone_data_time.csv')
fix_danger_zone_data('Amager_180min_geopop_div3/dijkstra-fastest-path-output/','180 min', 'time','danger_zone_data_time.csv')

```

```python
# ALGORITHMS
fix_danger_zone_data(folder_name='fastest-shortest-combined/amager-fastest-3-output/',column_to='Fastest', name='algorithms', remove_basis=True)
fix_danger_zone_data('fastest-shortest-combined/amager-shortest-3-output/','Shortest', 'algorithms', 'danger_zone_data_algorithms.csv')
fix_danger_zone_data('polaris-combined/amager-polaris-3-output/','Polaris', 'algorithms', 'danger_zone_data_algorithms.csv')
```

```python
# DIVERSIFYING ENDPOINTS
fix_danger_zone_data('fastest-shortest-combined/amager-fastest-1-output/','Fastest - div:1', 'diversifying', remove_basis=True)
fix_danger_zone_data('fastest-shortest-combined/amager-shortest-1-output/','Shortest - div:1', 'diversifying', original_df='danger_zone_data_diversifying.csv')
fix_danger_zone_data('polaris-combined/amager-polaris-1-output/','Polaris - div:1', 'diversifying', original_df='danger_zone_data_diversifying.csv')
# div 3
fix_danger_zone_data('fastest-shortest-combined/amager-fastest-3-output/','Fastest - div:3', 'diversifying',original_df='danger_zone_data_diversifying.csv')
fix_danger_zone_data('fastest-shortest-combined/amager-shortest-3-output/','Shortest - div:3', 'diversifying', original_df='danger_zone_data_diversifying.csv')
fix_danger_zone_data('polaris-combined/amager-polaris-3-output/','Polaris - div:3', 'diversifying', original_df='danger_zone_data_diversifying.csv')
```

```python
# Ravenna
fix_danger_zone_data('fastest-shortest-combined/ravenna-fastest-3-output/','Fastest', 'ravenna', remove_basis=True)
fix_danger_zone_data('fastest-shortest-combined/ravenna-shortest-3-output/','Shortest', 'ravenna', 'danger_zone_data_ravenna.csv')
fix_danger_zone_data('polaris-combined/ravenna-polaris-3-output/','Polaris', 'ravenna', 'danger_zone_data_ravenna.csv')
```
